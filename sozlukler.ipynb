```python
#soru3
kitap_stok = {"1984": 4, "Sefiller": 2, "Martı": 5}
kitap_adi = input('Kitap adı girin: ').lower()
adet = int(input('Kaç adet istediğinizi girin: '))

if kitap_adi in kitap_stok:
    if kitap_stok[kitap_adi] >= adet:
        print("siparişiniz tamamlandı.")
    else:
        print('stok yetersiz.')
else:
    print('geçersiz kitap adı.')
```

```python
#soru4
urunler = {
 "Telefon": {"yil": 2019},
 "Televizyon": {"yil": 2014},
 "Laptop": {"yil": 2021}
}

yeni_liste = []
for urun, bilgiler in urunler.items():
    if bilgiler['yil'] > 2015:
        yeni_liste.append(urun)
print("2015'ten sonra çıkan ürünler:", yeni_liste)        
```

```python
#soru8
sehir_bilgi = {
 "Berlin": {"nüfus": 3_600_000},
 "Madrid": {"nüfus": 3_200_000}
}

sehir_name = str(input('Şehir adı girin ')).lower()
bulundu = False
for sehir, nufus in sehir_bilgi.items():
    if sehir.lower() == sehir_name: 
        print(f"{sehir}: {nufus}")
        bulundu = True
        break

if not bulundu:
    print("Veri yok.")
```

```python
#soru 10
projeler = {
 "ProjeX": {"durum": "aktif", "süre": 3},
 "ProjeY": {"durum": "tamamlandı", "süre": 6},
 "ProjeZ": {"durum": "aktif", "süre": 2}
}
aktif_projeler= {}
for proje, bilgiler in projeler.items():
    if bilgiler["durum"] == "aktif":
        aktif_projeler[proje]= bilgiler
for proje, bilgiler in aktif_projeler.items():
    print(f"{proje}: {bilgiler}")
    
```
